```python
#  {RICHARD MUGO NJUGUNA-128100}


# TASK 1: NET SALARY CALCULATOR
# ZAWADI ENTERPRISES LTD

print("======================================")
print("   ZAWADI ENTERPRISES LTD")
print("       SALARY CALCULATOR")
print("======================================")

# Capture employee details
payroll_no = input("Enter Payroll Number: ")
name = input("Enter Employee Name: ")
gender = input("Enter Gender: ")
department = input("Enter Department: ")
basic_salary = float(input("Enter Basic Salary (Ksh): "))

# Calculate allowances
house_allowance = 6500
medical_allowance = 5500

# Calculate gross pay
gross_pay = basic_salary + house_allowance + medical_allowance

# Calculate PAYE according to the given table
if 40000 <= gross_pay <= 49000:
    paye = gross_pay * 0.06
elif 31000 <= gross_pay <= 39000:
    paye = gross_pay * 0.05
elif 20000 <= gross_pay <= 30000:
    paye = gross_pay * 0.04
elif 0 <= gross_pay <= 19000:
    paye = 0
else:
    paye = 0

# Calculate NHIF and NSSF
nhif = gross_pay * 0.02
nssf = gross_pay * 0.03

# Calculate total deductions
total_deductions = paye + nhif + nssf

# Calculate net pay
net_pay = gross_pay - total_deductions

# Display employee information and salary details
print("\n======================================")
print("       EMPLOYEE PAYSLIP")
print("======================================")

print("Payroll Number:", payroll_no)
print("Employee Name:", name)
print("Gender:", gender)
print("Department:", department)

print("--------------------------------------")
print("Basic Salary: Ksh", round(basic_salary, 2))
print("House Allowance: Ksh", house_allowance)
print("Medical Allowance: Ksh", medical_allowance)
print("Gross Pay: Ksh", round(gross_pay, 2))

print("--------------------------------------")
print("P.A.Y.E: Ksh", round(paye, 2))
print("N.H.I.F: Ksh", round(nhif, 2))
print("N.S.S.F: Ksh", round(nssf, 2))
print("Total Deductions: Ksh", round(total_deductions, 2))

print("--------------------------------------")
print("NET PAY: Ksh", round(net_pay, 2))
print("======================================")
```

```python

# TASK 2: DISEASE DIAGNOSIS PROGRAM

      # AMANI MEDICAL CENTRE

print("======================================")
print(" WELCOME TO AMANI MEDICAL CENTRE")
print("======================================")

# Capture patient details
name = input("Enter Patient's Name: ")
gender = input("Enter Gender: ")
age = int(input("Enter Age: "))
residence = input("Enter Place of Residence: ")

# Capture two symptoms
print("\nEnter the patient's symptoms")

symptom1 = input("Enter Symptom 1: ").strip().lower()
symptom2 = input("Enter Symptom 2: ").strip().lower()

# Determine possible diagnosis & Both symptoms must match the programmed combination

if (symptom1 == "vomiting" and symptom2 == "diarrhea") or \
   (symptom1 == "diarrhea" and symptom2 == "vomiting"):
    diagnosis = "Typhoid"

elif (symptom1 == "vomiting" and symptom2 == "fever") or \
     (symptom1 == "fever" and symptom2 == "vomiting"):
    diagnosis = "Malaria"

elif (symptom1 == "chest pain" and symptom2 == "wheezing sound") or \
     (symptom1 == "wheezing sound" and symptom2 == "chest pain"):
    diagnosis = "Pneumonia"

elif (symptom1 == "frequent urination" and
      symptom2 == "fluctuations of sugar levels") or \
     (symptom1 == "fluctuations of sugar levels" and
      symptom2 == "frequent urination"):
    diagnosis = "Diabetes"

else:
    diagnosis = "No matching diagnosis found"

# Display patient information and diagnosis
print("\n======================================")
print("       PATIENT DIAGNOSIS REPORT")
print("======================================")

print("Patient Name:", name)
print("Gender:", gender)
print("Age:", age)
print("Place of Residence:", residence)

print("--------------------------------------")
print("Symptom 1:", symptom1.title())
print("Symptom 2:", symptom2.title())

print("--------------------------------------")
print("Possible Diagnosis:", diagnosis)

print("--------------------------------------")

if diagnosis == "No matching diagnosis found":
    print("Please consult a qualified medical professional.")
else:
    print("Please consult a qualified medical professional")
    print("for proper medical examination and confirmation.")

print("======================================")
```
